# BAWE Corpus — Distribution Analysis\n\nTwo distributions from the `BAWE_by_student/` folder:\n1. **Essays per student** — how many essays each of the 627 students contributed\n2. **Words per essay** — length distribution across all essays

```python
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np

STUDENT_DIR = Path("BAWE_by_student")

# Load data
essays_per_student = []
words_per_essay = []

for f in sorted(STUDENT_DIR.glob("*.txt")):
    lines = f.read_text(encoding="utf-8").splitlines()
    lines = [l for l in lines if l.strip()]   # drop blank lines
    essays_per_student.append(len(lines))
    for line in lines:
        words_per_essay.append(len(line.split()))

print(f"Students : {len(essays_per_student)}")
print(f"Essays   : {sum(essays_per_student)}")
print(f"Essays/student — min {min(essays_per_student)}, max {max(essays_per_student)}, mean {np.mean(essays_per_student):.1f}")
print(f"Words/essay    — min {min(words_per_essay)}, max {max(words_per_essay)}, mean {np.mean(words_per_essay):.0f}")
```

## 1. Distribution of Essays per Student

```python
fig, ax = plt.subplots(figsize=(9, 4))

counts, bins = np.unique(essays_per_student, return_counts=True)
ax.bar(counts, bins, width=0.7, color="steelblue", edgecolor="white")

ax.set_xlabel("Number of essays", fontsize=12)
ax.set_ylabel("Number of students", fontsize=12)
ax.set_title("Distribution of Essays per Student (n=627 students)", fontsize=13)
ax.xaxis.set_major_locator(ticker.MaxNLocator(integer=True))
ax.yaxis.set_major_locator(ticker.MaxNLocator(integer=True))

# annotate median and mean
med = np.median(essays_per_student)
mean = np.mean(essays_per_student)
ax.axvline(med,  color="darkorange", linestyle="--", linewidth=1.5, label=f"Median = {med:.0f}")
ax.axvline(mean, color="crimson",    linestyle=":",  linewidth=1.5, label=f"Mean = {mean:.1f}")
ax.legend()

plt.tight_layout()
plt.show()
```

## 2. Distribution of Words per Essay

```python
fig, ax = plt.subplots(figsize=(9, 4))

ax.hist(words_per_essay, bins=60, color="teal", edgecolor="white", alpha=0.85)

ax.set_xlabel("Words per essay", fontsize=12)
ax.set_ylabel("Number of essays", fontsize=12)
ax.set_title(f"Distribution of Words per Essay (n={len(words_per_essay):,} essays)", fontsize=13)

med  = np.median(words_per_essay)
mean = np.mean(words_per_essay)
ax.axvline(med,  color="darkorange", linestyle="--", linewidth=1.5, label=f"Median = {med:,.0f}")
ax.axvline(mean, color="crimson",    linestyle=":",  linewidth=1.5, label=f"Mean = {mean:,.0f}")
ax.legend()

plt.tight_layout()
plt.show()
```
